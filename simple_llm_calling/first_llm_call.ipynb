```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")
response=llm.invoke("Who is the father of India? ")

print(response.text)
```

```python
#for creativity use the parameter temperature as 0
#CREATIVE means choose the least probability next token

llm=ChatGoogleGenerativeAI(model="gemma-4-31b-it",temperature=1);

prompt="Suggest me a name for my new born son"

response=llm.invoke(prompt)

print(response.text)
```
