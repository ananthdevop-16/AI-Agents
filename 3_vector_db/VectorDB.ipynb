```python
import chromadb

client=chromadb.Client()


```

```python
#its like giving name to a table in db
collection=client.create_collection(
    name="art-ic-les"
)
```

```python
collection.add(
    ids=["id1","id2","id3","id4"],
    documents=[
        "Apple is leading in a smart phone game with iPhone sales up by 35%",
        "Tesla booked a minor profit of 1 billion $ of Q2",
        "Apples are high in fiber, vitamin C, and antioxidants",
        "SpaceX got NASA contract worth oh 10 billion $"
    ]
)
```

```python
#fetch documents and its meanings(embeddings)
data =collection.get(
    include=["embeddings","documents","metadatas"]
)

#print embeddings
for i,emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}")
    print("Text : ",data["documents"][i])
    print("Embeddind length : ",len(emb))
    print("First 10 values : ",emb[:10])
```

```python

#queries
results=collection.query(
    query_texts=["This is a query related to Elon musk"],
    n_results=2
)

#distance represents how the result is distant from the query
print(results)
```
